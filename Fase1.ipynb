```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('steam_games.csv')
df_ale = pd.read_csv('steam_games.csv').head(20)
```

```python
jogos_gratis = (df['Price'] == 0).sum()
jogos_pagos = (df['Price'] > 0).sum()
        
plt.figure(figsize=(6, 6))
plt.pie(
[jogos_gratis, jogos_pagos],
labels=['Gratuitos', 'Pagos'],
autopct='%1.1f%%',
colors=['lightblue', 'orange'],
startangle=140
)
plt.title("Percentual de Jogos Gratuitos vs Pagos")
plt.show()
```

```python
df['Release date'] = pd.to_datetime(df['Release date'], errors='coerce')
df['Ano'] = df['Release date'].dt.year
contagem_anos = df['Ano'].value_counts()
max_lancamentos = contagem_anos.max()
        
plt.figure(figsize=(10, 5))
sns.barplot(x=contagem_anos.index, y=contagem_anos.values, color='royalblue')
plt.title("Lançamentos de Jogos por Ano")
plt.xlabel("Ano")
plt.ylabel("Quantidade de Jogos")
plt.xticks(rotation=90)
plt.show()
```

```python
total_jogos = len(df)
compatibilidade_windows = df['Windows'].sum()
compatibilidade_mac = df['Mac'].sum()
compatibilidade_linux = df['Linux'].sum()
percentual_so = {
'Windows': (compatibilidade_windows / total_jogos) * 100,
'Mac': (compatibilidade_mac / total_jogos) * 100,
'Linux': (compatibilidade_linux / total_jogos) * 100
}
        
plt.figure(figsize=(6, 6))
plt.pie(
percentual_so.values(),
labels=percentual_so.keys(),
autopct='%1.1f%%',
colors=['royalblue', 'gold', 'red'],
startangle=140
)
plt.title("Compatibilidade com Sistemas Operacionais")
plt.show()
```
